# Set up ZCU216 for RFSoC MTS

This overlay uses the ZCU216, CLK104 clock card, and XM655 analogue breakout.
Use the ZCU216 PYNQ 4.0 image and install the repository with its ZCU216 installer.

## Check the image and driver

The ZCU216 must identify itself as `ZCU216`. The patched `xrfdc` package provides
the MTS methods and the installer supplies the 500 MHz CLK104 register profiles.

```python
from importlib.metadata import version
from pathlib import Path

import xrfdc

board = Path('/proc/device-tree/chosen/pynq_board').read_bytes().rstrip(b'\0').decode()
assert board == 'ZCU216', f'Expected ZCU216, found {board}'
assert version('pynq') == '4.0.0'

methods = ('adc_mts_init', 'dac_mts_init', 'adc_mts_sync', 'dac_mts_sync',
           'mts_sysref_enable')
missing = [name for name in methods if not hasattr(xrfdc.RFdc, name)]
assert not missing, f'Missing MTS methods: {missing}'
print(f'{board} · PYNQ {version("pynq")} · MTS driver available')
```

## Connect one loopback

With the player idle, connect DAC tile 228 channel 0 to ADC tile 224 channel 0
with the appropriate DC block and attenuation. That ADC is `adc_00` in the
demonstrator. Select the low-frequency balun path for the default 200 MHz NCO.
Start with this single loopback before adding channels.
